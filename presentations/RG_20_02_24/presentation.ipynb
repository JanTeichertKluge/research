## Motivating Example - PLR

#### Partially linear regression model (PLR) 

$$
\begin{align}
Y &= \theta_0 D + g_0(X) + \varepsilon, & \mathbb{E}[\varepsilon | X, D] = 0 \label{eq:plr1} \\
D &= m_0(X) + \vartheta, & \mathbb{E}[\vartheta | X] = 0 \label{eq:plr2}
\end{align}
$$

with

::: {.column width="50%"}

- Output $Y$
- Treatment $D$
- Additional features $X = (X_1, \dots)$
- Stochastic errors $\varepsilon$ and $V$

## Motivating Example - DML

#### Double/Debiased Machine Learning (DML)

- Use machine learning methods to fit 
$$
\begin{align}
l_0(X) &:= \mathbb{E}[Y | X] \label{eq:l_hat} \\
m_0(X) &:= \mathbb{E}[D | X] \label{eq:m_hat}
\end{align}
$$
- Construct the orthogonalized score (either cross-fitting or split the sample) for a fitted nuisance learner $\hat{\eta}_0:=(\hat{l}_0, \hat{m}_0)$
$$
\begin{align}
\psi(W, \theta, \hat{\eta}) := \left(Y - \hat{l}(X) - \theta\left(D - \hat{m}(X)\right)\right) \cdot \left(D - \hat{m}(X)\right)
\end{align}
$$
- Compute the estimate as the solution of 
$$
0 = \frac{1}{n}\sum_{i=1}^n \psi(W, \hat{\theta}, \hat{\eta}_0)
$$

::: {.notes}
- We start by using machine learning to fit functions that capture the conditional expectations of the Output and the treatment Variable given our (unstructured and high dimensional) set of covariates:

\begin{align*}
l_0(X) &:= \mathbb{E}[\ln(Q)|X] \\
m_0(X) &:= \mathbb{E}[\ln(P)|X]
\end{align*}

- Next, we construct an orthogonalized score for a fitted nuisance learner denoted as $\hat{\eta}_0$, which comprises $\hat{l}_0$ and $\hat{m}_0$.

$$
\begin{align}
\psi(W, \theta, \hat{\eta}_0) := \left(\ln(Q) - \hat{l}_0(X) - \theta\left(\ln(P) - \hat{m}_0(X)\right)\right)\left(\ln(P) - \hat{m}_0(X)\right)
\end{align}
$$

- On the final step, we calculate the estimate by solving the following equation:

$$
0 = \frac{1}{n}\sum_{i=1}^n \psi(W, \hat{\theta}_0, \hat{\eta}_0)
$$
:::


## Motivating Example - DML

<br>

- Assumptions for the nuisance elements (see @dml2018)
$$
\begin{align}
\lVert \hat{m}(X) - m_0(X) \rVert_{P,2} \cdot \left(\lVert \hat{m}(X) - m_0(X) \rVert_{P,2} +  \lVert \hat{l}(X) - l_0(X)\rVert _{P,2}\right) \le \delta_N N^{-1/2}.
\end{align}
$$

- Under some regularity conditions the estimator $\hat{\theta}$ concentrates in $1/\sqrt{n}$-neighborhood of $\theta_0$ and
$$
\sqrt{n}(\hat{\theta} - \theta_0) \to \mathcal{N}(0, \sigma^2)
$$

. . .

$\Rightarrow$ simple approach to control for confounders, *if they can be used for ML models!*

::: {.notes}
- We rely on the approach of @dml2018

**Nuisance Element Assumptions:** We assume that the difference between our estimated nuisance functions ($\hat{m}_0$ and $\hat{l}_0$) and the true functions ($m_0$ and $\ell_0$) is bounded by a factor scaling with the sample size, as given by the equation:

    $$
    \lVert \hat{m}_0 - m_0 \rVert_{P,2} \times \left(\lVert \hat{m}_0 - m_0 \rVert_{P,2} +  \lVert \hat{\ell}_0 - \ell_0\rVert _{P,2}\right) \le \delta_N N^{-1/2}
    $$

Under these assumption and some regularity conditions, our estimator $\hat{\theta}_0$ converges to a neighborhood of the true parameter $\theta_0$ at a rate of $1/\sqrt{n}$. Furthermore, the square root of the sample size times the difference between our estimated parameter and the true parameter converges to a normal distribution:

    $$
    \sqrt{n}(\hat{\theta}_0 - \theta_0) \to \mathcal{N}(0, \sigma^2)
    $$

  - This approach implemented in the DoubleML package can control for product features, if they can be used in (std.) ML models

## DoubleML Deep - More Details

<br>

- Use pretrained models for unstructured data e.g. Bert and Beit

- Basic models for tabular data seem sufficient

- Monitor the nuisance losses
  - $\lVert \hat{l}_0 - \ln(Q) \rVert_{P,2} \le \lVert \hat{l}_0 -l_0 \rVert_{P,2} + \sigma_{\tilde{\epsilon}}$
  - $\lVert \hat{m}_0 - \ln(P) \rVert_{P,2} \le \lVert \hat{m}_0 - m_0 \rVert_{P,2} + \sigma_V$

::: {.notes}
- **Unstructured Data Models:** We employ pre-trained models like BERT and BEiT to extract valuable information from unstructured data, including product descriptions and images.

- **Tabular Data Models:** Basic models for tabular data have proven to be effective.

- **Monitoring Nuisance Losses:** To ensure the reliability of our estimations, we diligently monitor nuisance losses. The following conditions are crucial for assessing the quality of our models:

  - $\lVert Y - \hat{l}(X) \rVert_{P,2} &\le \lVert l_0 - \hat{l}(X) \rVert_{P,2} + \sigma_{\epsilon}$ 

  - $ \lVert D - \hat{m}(X) \rVert_{P,2} &\le \lVert m_0(X) - \hat{m}(X)\rVert_{P,2} + \sigma_{\vartheta}$

  These conditions help maintain the validity of our estimates by ensuring that the discrepancies between our estimated nuisance functions and their true counterparts, along with stochastic errors, remain within acceptable bounds.
:::



## Amazon Toys Dataset
:::: {.columns .my-table}

::: {.column width="50%"}

**Continous Variables**

  * Reviews: Rating                                   
  * Reviews: Review Count                             
  * New Offer Count: Current                          
  * Count of retrieved live offers: New, FBA          
  * Count of retrieved live offers: New, FBM     

:::

::: {.column width="50%"}
**Categorical Variables**

  * Lightning Deals: Upcoming Deal  
  * Buy Box: Is FBA      

:::

::::


Scatterplots with OLS Regression of Tabular Variables and $\ln(Q)$
![](figures/tab_pairplot.png)

::: {.notes}
- In addition to the primary variables we discussed, the Amazon Toys Dataset includes several other variables, both continuous and categorical, providing further context and insight into products in this category:

  **Continuous Variables:**
  - **Reviews: Rating:** Reflects the average rating of the product based on customer reviews, serving as an indicator of customer satisfaction and product quality.
  - **Reviews: Review Count:** The total count of reviews for the product, offering insights into the level of customer engagement and feedback.
  - **New Offer Count: Current:** Indicates the number of new offers currently available for the product, reflecting market competition.
  - **Count of Retrieved Live Offers: New, FBA:** The count of retrieved live offers for new products that are fulfilled by Amazon (FBA), which can affect pricing and demand.
  - **Count of Retrieved Live Offers: New, FBM:** Similar to the previous variable, but for new products fulfilled by merchants (FBM).

  **Categorical Variables:**
  - **Lightning Deals: Upcoming Deal:** Indicates whether the product is associated with an upcoming lightning deal, which can impact pricing dynamics.
  - **Buy Box: Is FBA:** A categorical variable identifying whether the product is associated with the "Buy Box" when fulfilled by Amazon (FBA).

- The provided scatterplot of tabular variables and the natural logarithm of demand ($\ln(Q)$) offers a visual representation of the relationships between these variables, aiding our understanding of their interactions and potential influences on demand.
:::

## Amazon Toys Dataset
#### Data Examples

:::: {.columns}

::: {.column width="50%"}

Kinsmart Set of 4 McLaren 720s Toy | ...
![](figures/img0.png)

:::

::: {.column width="50%"}

Pixar Cars Mack Uncle Lightning...
![](figures/img1.png)

:::

::::

::: {.notes}

:::

## Baseline


:::: {.columns}

::: {.column width="45%"}

* Baseline estimate with tabular covariates, OLS model for regression 
  - $\ln(Q) = \theta_0 \ln(P) + \beta^T X + \epsilon$
    - $R^2 = 0.330$

| 2.5 %   | $\theta$ | 97.5 % |
| --------   | ------- | ------- |
| -0.072 | -0.046 | -0.019 |

:::

::: {.column width="55%"}

![Centered logarithmic negative sales rank over centered logarithmic price as binned plot](figures/baseline_binned_plot.png)

:::

::::

::: {.notes}
- In our initial baseline estimate, we begin with a simple OLS (Ordinary Least Squares) regression model that does not include covariates ($X). This baseline model serves as a reference point for comparison.

- The baseline model is formulated as follows:

  $$
  \ln(Q) = \beta_0 + \beta_1 \ln(P) + \epsilon
  $$

  Where:
  - $\ln(Q)$ represents the natural logarithm of demand.
  - $\ln(P)$ represents the natural logarithm of price.
  - $\beta_0$ and $\beta_1$ are regression coefficients.
  - $\epsilon$ denotes the error term.

- The $R^2$ value for this baseline model, indicating the proportion of variability in demand explained by price, is exceptionally low at $0.000$. This implies that the model, without considering covariates, has minimal explanatory power.

- The plot on the right illustrates the centered logarithmic negative sales rank over the centered logarithmic price as a binned plot, helping visualize the relationship between sales rank and price within this baseline model. It becomes evident that the relationship is far from straightforward and requires more advanced modeling techniques to capture the data's complexity.

- The 95% confidence interval for $\beta_1$ is presented below:

  - 2.5%: -0.011
  - Median: 0.0203
  - 97.5%: 0.052

  This baseline serves as the starting point for understanding the relationship between price and demand.
:::

## Baseline

:::: {.columns}

::: {.column width="45%"}

* Estimate with tabular data as covariates $X$ in $\texttt{DoubleMLPLR}$ model with $\texttt{RandomForest}$ regressors
  - $l_0(X) := \mathbb{E}[\ln(Q)|X]$
    - $R^2_{l_0} = 0.5986$
  - $m_0(X) := \mathbb{E}[\ln(P)|X]$
    - $R^2_{m_0} = 0.1884$

| 2.5 %   | $\theta$ | 97.5 % |
| --------   | ------- | ------- |
| -0.132 | -0.1098 | -0.080 |

:::

::: {.column width="55%"}

![Residualised logarithmic negative sales rank over residualised logarithmic price as binned plot](figures/baseline_dmlplr.png)

:::

::::

::: {.notes}
The binned plot on the right illustrates the relationship between residualized logarithmic negative sales rank and residualized logarithmic price, confirming the improvements in model performance. The residuals indicate that a significant portion of the variability in demand and price has been accounted for by this model, allowing for more accurate and reliable estimations of price elasticities.

- The 95% confidence interval for $\theta_0$ is presented below:

  - 2.5%: -0.132
  - Median: -0.1098
  - 97.5%: -0.080

  This refined baseline serves as a crucial benchmark for evaluating the performance of the more advanced DoubleML Deep model, which further leverages unstructured data to enhance price elasticity estimation.
:::


## Deep Learning Models

* In this study, a variety of input data combinations were used to train $\texttt{DoubleMLPLRDeep}$ models [^top]:

| Key | Used Confounders |
| --------   | ------- |
| $\texttt{img}$ | $X = (X_{img})$ |
| $\texttt{imgtab}$ | $X = (X_{img}, X_{tab})$ |
| $\texttt{txtimg}$ | $X = (X_{txt}, X_{img})$ |
| $\texttt{txt}$ | $X = (X_{txt})$ |
| $\texttt{txttab}$ | $X = (X_{txt}, X_{tab})$ |
| $\texttt{txtimgtab}$ | $X = (X_{txt}, X_{img}, X_{tab})$ |


. . .

$\Rightarrow$ The aim is to emphasize the benefits of utilizing unstructured data.

[^top]: Subsequently, the top 5 performing instances for each combination were evaluated based on the minimum RMSE score of the validation data.


## Deep Learning Models: RMSE-Scores

:::: {.columns}

::: {.column width="50%"}

- Combined RMSE Score (Training)

![](figures/train_comb_loss_mod.png)


:::

::: {.column width="50%"}

- Combined RMSE Score (Validation)

![](figures/val_comb_loss_mod.png)


:::

::::

::: {.notes}

:::

## Deep Learning Models: $R^2$-Scores

:::: {.columns}

::: {.column width="50%"}

- $R^2$ of log(Quantity) on Validation Set

![](figures/R2_ml_l_mod.png)


:::

::: {.column width="50%"}

- $R^2$ of log(Price) on Validation Set

![](figures/R2_ml_m_mod.png)


:::

::::

::: {.notes}

:::

## First Results


:::: {.columns}

::: {.column width="50%"}

* Using different combinations of tabular, image and text data we obtain the following estimates

![](figures/theta_ln_p.png)
:::

::: {.column width="50%"}

![Continous Estimates from $\texttt{DoubleMLPLR}$ on Validation Set during the Training Process](figures/cont_dml_sub.png)

:::

::::


::: {.notes}
- The visual representation shows continuous estimates from the `DoubleMLPLR` model during the training process, which converge to stable estimates.
:::


## Comparison to Baseline Estimates
| Model | Covariates | $R^2_{l_0}$ |$R^2_{m_0}$ | $\hat{\theta}_0$
| --- | --- | --- | --- | --- |
| $\texttt{OLS}$ | $X=X_{tab}$ | $0.3300$ | - | $-0.0455$ |
| $\texttt{DoubleMLPLR}$ | $X=X_{tab}$ | $0.5986$ | $0.1884$ | $-0.1098$ |
| $\texttt{DoubleMLPLRDeep}$[^1] | $X=(X_{tab}, X_{txt}, X_{img})$ | $0.5990$	 | $0.6765$ | $-0.2794$|

<center>
![](figures/baseline_vs_deep.png){width=800}
</center>

[^1]: Mean values of the $\texttt{DoubleMLDeep}$ Model with tabular, text and image data as covariates.

::: {.notes}
When comparing the results, it is noticeable that the predictive performance, measured by the R-squared value, does not really improve, at least for the sales variable. However, the predictive performance for the price is significantly improved.
When looking at the estimators for price elasticity, it is noticeable that the estimate by DoubleMLDeep has the highest absolute value, while the OLS estimator delivers the lowest absolute value. 
:::

## Future Research Direction

<br>

- Semi-synthetical data for further evaluation

- Further applications (We are really looking forward to possible collaborations)

- Extensions of the [DoubleML package](https://docs.doubleml.org/stable/index.html)

- Using multiple time periods


::: {.notes}

:::


# References

## References

::: {#refs}
:::


## Affiliations

<div class="affiliation-grid">
<div class="affiliation">
<a href="https://www.bwl.uni-hamburg.de/en/statistik/allgemein.html"><img src="../shared_files/uhh_logo.png" alt="UHH" height="150px"></a>
<p>University of Hamburg</p>
<p><small>https://www.bwl.uni-hamburg.de/en/statistik/allgemein.html</small></p>
</div>
<div class="affiliation">
<a href="https://stat.mit.edu/about/"><img src="../shared_files/mit_logo.png" alt="MIT" height="150px"></a>
<p>Massachusetts Institute of Technology</p>
<p><small>https://stat.mit.edu/about/</small></p>
</div>
<div class="affiliation">
<a href="https://economicai.com/"><img src="../shared_files/logo_eai_full.png" alt="EAI" height="100px"></a>
<p>Economic AI</p>
<p><small>https://economicai.com/</small></p>
</div>
<div class="affiliation">
<a href="https://docs.doubleml.org/stable/index.html"><img src="../shared_files/logo.png" alt="DML" height="100px"></a>
<p>DoubleML</p>
<p><small>https://docs.doubleml.org/stable/index.html</small></p>
</div>
</div>

. . .

<div style="text-align: center">
$\Rightarrow$ **Thanks for your Attention!**
</div>

::: {.notes}
- Economic AI: for industry collaborations and trainings on doubleml please check out the website of economicai.com ! 
:::
